```python
import os
from rembg import remove
from PIL import Image
from google.colab import drive

# Montar Google Drive
drive.mount('/content/drive')

# Definir las rutas base
carpeta_entrada_base = '/content/drive/My Drive/papa_nativa_24_2'
carpeta_salida_base = '/content/drive/My Drive/unidad/imagenes'

# Crear carpeta de salida base si no existe
if not os.path.exists(carpeta_salida_base):
    os.makedirs(carpeta_salida_base)

# Función para crear un marco blanco de la imagen original
def crear_marco(img):
    # Crear una imagen blanca del mismo tamaño
    white_image = Image.new("RGB", img.size, (255, 255, 255))
    return white_image

# Función para procesar cada imagen, eliminar el fondo y combinarla con el marco
def procesar_imagen(nombre_archivo, carpeta_entrada, carpeta_salida, carpeta_num, j):
    # Crear carpeta de salida si no existe
    os.makedirs(carpeta_salida, exist_ok=True)

    input_path = os.path.join(carpeta_entrada, nombre_archivo)
    temp_path = 'temporal.png'  # Guardar la imagen sin fondo como 'temporal.png' en el directorio actual
    marco_path = 'marco.png'
    output_path = os.path.join(carpeta_salida, f'{carpeta_num} ({j}).jpg')  # Usar carpeta_num y j para el nombre del archivo

    # Eliminar fondo
    with open(input_path, 'rb') as input_file:
        with open(temp_path, 'wb') as temp_file:
            input_image = input_file.read()
            output_image = remove(input_image)
            temp_file.write(output_image)

    print(f"Imagen sin fondo guardada temporalmente como {temp_path}")

    # Pegar imagen sin fondo sobre un nuevo marco blanco
    with Image.open(temp_path) as foreground:
        marco = crear_marco(foreground)  # Crear un nuevo marco para cada imagen
        foreground = foreground.resize(marco.size)  # Redimensionar si es necesario
        marco.paste(foreground, (0, 0), foreground)  # Usar la imagen sin fondo como máscara
        marco.save(output_path)

    print(f"Imagen final guardada en {output_path}")

# Recorrer cada carpeta (1 a 84) en la carpeta base de entrada
for carpeta_num in range(75, 85):
    carpeta_entrada = os.path.join(carpeta_entrada_base, str(carpeta_num))
    carpeta_salida = os.path.join(carpeta_salida_base, str(carpeta_num))

    # Listar archivos en la carpeta
    archivos = [archivo for archivo in os.listdir(carpeta_entrada) if archivo.endswith('.jpg') or archivo.endswith('.png')]
    num_archivos = len(archivos)
    j = 0
    print(f'Carpeta {carpeta_num}: {num_archivos} archivos encontrados')

    # Procesar cada imagen en la carpeta actual
    for nombre_archivo in archivos:  # Recorre todos los archivos de imagen
        procesar_imagen(nombre_archivo, carpeta_entrada, carpeta_salida, carpeta_num, j + 1)
        j += 1
```
