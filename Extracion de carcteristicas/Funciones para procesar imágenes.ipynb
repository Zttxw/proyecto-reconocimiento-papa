```python
import os
import cv2
import numpy as np
from PIL import Image

# 1. Función para eliminar el fondo de una imagen
def remove_background(image_path):
    """
    Elimina el fondo de una imagen y lo reemplaza con transparencia.
    Requiere OpenCV para manejar la eliminación del fondo.
    """
    image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Aplicar umbral adaptativo para crear una máscara
    _, mask = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    # Crear un canal alfa basado en la máscara
    bg_removed = cv2.cvtColor(image, cv2.COLOR_BGR2BGRA)
    bg_removed[:, :, 3] = mask

    # Convertir la imagen a formato PIL para mayor flexibilidad
    return Image.fromarray(cv2.cvtColor(bg_removed, cv2.COLOR_BGRA2RGBA))
```

```python
# 2. Función para redimensionar la imagen
def resize_image(image, size=(128, 128)):
    """
    Redimensiona una imagen a un tamaño especificado.
    """
    return image.resize(size, Image.LANCZOS)
```

```python
# 3. Función para agregar un fondo blanco
def add_white_background(image):
    """
    Combina la imagen con transparencia (RGBA) con un fondo blanco sólido.
    """
    # Crear un fondo blanco del mismo tamaño
    background = Image.new("RGB", image.size, (255, 255, 255))
    image_no_alpha = Image.alpha_composite(background.convert("RGBA"), image)
    return image_no_alpha.convert("RGB")
```

```python
# 4. Función para guardar la imagen
def save_image(image, save_path):
    """
    Guarda la imagen en el directorio especificado.
    """
    image.save(save_path, format="PNG")
```

```python
# 5. Función principal para procesar imágenes en un directorio
def process_images(input_folder, output_folder, size=(128, 128)):
    """
    Procesa todas las imágenes en una carpeta:
    - Elimina el fondo.
    - Redimensiona las imágenes.
    - Agrega fondo blanco.
    - Guarda las imágenes procesadas.
    """
    os.makedirs(output_folder, exist_ok=True)  # Crear carpeta si no existe

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):  # Validar formato
            image_path = os.path.join(input_folder, filename)
            print(f"Procesando: {filename}")

            # Eliminar fondo
            try:
                image = remove_background(image_path)
            except Exception as e:
                print(f"Error al eliminar fondo para {filename}: {e}")
                continue

            # Redimensionar imagen
            image = resize_image(image, size)

            # Agregar fondo blanco
            image = add_white_background(image)

            # Guardar imagen procesada
            save_path = os.path.join(output_folder, filename)
            save_image(image, save_path)
            print(f"Guardada imagen procesada en: {save_path}")
```
